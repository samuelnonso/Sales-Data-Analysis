### Most Selling Product

<h4> Most selling product by quantity</h4>

```python
 df.head()
```

```python
df.groupby('Product')['Quantity'].sum()
```

```python
df.groupby('Product')['Quantity'].sum().sort_values(ascending=True)
```

```python
most_quantity = df.groupby('Product')['Quantity'].sum().sort_values(ascending=True)
most_quantity
```

```python
type(most_quantity)                  #to know the type data
```

```python
#since its in a series fromat, we need to chage it to a dataframe  format 

most_quantity = most_quantity.reset_index()
most_quantity
```

```python
type(most_quantity)            # to know the datatype
```

```python
# plotting the data
plt.figure(figsize=(9,4))
plt.bar(most_quantity['Product'], most_quantity['Quantity'], color=['red', 'black', 'green', 'cyan', 'orange'], width=0.5)
plt.title("Most Selling Product - By Quantity")
plt.xlabel('Product')
plt.ylabel('Quantity')
plt.show()
```

<h4> Most selling product By Revenue </h4>
